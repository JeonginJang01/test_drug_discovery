```python
# 단일 체인 분리 (A 만 남기기)
awk 'BEGIN{OFS=""}
  /^ATOM|^HETATM/ { if (substr($0,22,1)=="A") print; next }
  /^REMARK|^TER|^END|^MODEL|^ENDMDL/ { print }
' /home/jeongin/eupatilin/data/protein/GSK3/6hk4/pdb/6HK4.pdb > /home/jeongin/eupatilin/data/protein/GSK3/6hk4/pdb/6HK4_A.pdb
```

```python
# A, B 체인 남기기 (B 체인이 알파체인으로 붙어있을 경우)
awk 'BEGIN{OFS=""}
  /^ATOM|^HETATM/ {
      c = substr($0,22,1)
      if (c=="A" || c=="B") print
      next
  }
  /^REMARK|^TER|^END|^MODEL|^ENDMDL/ { print }
' /home/jeongin/eupatilin/data/protein/PPAR_gamma/pdb/3KMG.pdb > /home/jeongin/eupatilin/data/protein/PPAR_gamma/chain_split/3KMG.pdb
```
